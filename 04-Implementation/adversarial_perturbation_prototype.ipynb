```python
from pathlib import Path

import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt
```

```python
BASE_DIR = Path(
    r"C:\Users\admin\Desktop\HWU\ResearchMethods\MachineLearningCVE"
)

DATA_FILE = BASE_DIR / "processed" / "cicids2017_clean.csv"
MODEL_FILE = BASE_DIR / "processed" / "random_forest_pilot.joblib"

data = pd.read_csv(DATA_FILE)
rf = joblib.load(MODEL_FILE)

print("Dataset:", data.shape)
print("Model loaded.")
```

```python
feature_columns = [
    col for col in data.columns
    if col not in ["Label", "BinaryLabel", "SourceFile"]
]

print("Number of features:", len(feature_columns))
```

```python
attack_data = data[data["BinaryLabel"] == 1].sample(
    n=20_000,
    random_state=42
)

X_attack = attack_data[feature_columns]

baseline_pred = rf.predict(X_attack)

baseline_detected = baseline_pred == 1

print("Attack sample:", len(X_attack))
print("Correctly detected before perturbation:",
      baseline_detected.sum())
print("Baseline detection rate:",
      baseline_detected.mean())
```

```python
X_detected = X_attack.loc[baseline_detected].copy()

print("Flows used for evasion test:",
      len(X_detected))
```

```python
def apply_timing_perturbation(X_input, factor):
    X_mod = X_input.copy()

    timing_features = [
        "Flow Duration",
        "Flow IAT Mean",
        "Flow IAT Std",
        "Flow IAT Max",
        "Flow IAT Min",

        "Fwd IAT Total",
        "Fwd IAT Mean",
        "Fwd IAT Std",
        "Fwd IAT Max",
        "Fwd IAT Min",

        "Bwd IAT Total",
        "Bwd IAT Mean",
        "Bwd IAT Std",
        "Bwd IAT Max",
        "Bwd IAT Min"
    ]

    rate_features = [
        "Flow Bytes/s",
        "Flow Packets/s",
        "Fwd Packets/s",
        "Bwd Packets/s"
    ]

    for feature in timing_features:
        if feature in X_mod.columns:
            X_mod[feature] = X_mod[feature] * factor

    for feature in rate_features:
        if feature in X_mod.columns:
            X_mod[feature] = X_mod[feature] / factor

    return X_mod
```

```python
factors = [1.00, 1.05, 1.10, 1.20, 1.30]

results = []

for factor in factors:

    X_modified = apply_timing_perturbation(
        X_detected,
        factor
    )

    predictions = rf.predict(X_modified)

    detection_rate = (predictions == 1).mean()
    evasion_rate = (predictions == 0).mean()

    results.append({
        "TimingFactor": factor,
        "DetectionRate": detection_rate,
        "EvasionRate": evasion_rate
    })

results_df = pd.DataFrame(results)

results_df
```

```python
plt.figure(figsize=(7, 4))

plt.plot(
    results_df["TimingFactor"],
    results_df["DetectionRate"],
    marker="o"
)

plt.xlabel("Timing perturbation factor")
plt.ylabel("Detection rate")
plt.title("Preliminary Constraint-Aware Timing Perturbation")

plt.ylim(0, 1.02)
plt.grid(True)

plt.show()
```

```python
def apply_unconstrained_perturbation(
    X_input,
    strength,
    random_state=42
):
    """
    Preliminary unconstrained feature-space perturbation.

    Selected numerical features are changed independently.
    No dependencies between features are enforced.

    This is a feasibility prototype, not a final attack algorithm.
    """

    rng = np.random.default_rng(random_state)

    X_mod = X_input.copy()

    unconstrained_features = [
        "Flow Duration",
        "Flow Bytes/s",
        "Flow Packets/s",
        "Flow IAT Mean",
        "Flow IAT Std",
        "Fwd Packets/s",
        "Bwd Packets/s",
        "Packet Length Mean",
        "Average Packet Size"
    ]

    for feature in unconstrained_features:

        if feature not in X_mod.columns:
            continue

        # Independent multiplicative change for every row
        noise = rng.uniform(
            low=-strength,
            high=strength,
            size=len(X_mod)
        )

        X_mod[feature] = (
            X_mod[feature] * (1 + noise)
        )

        # Prevent negative values
        X_mod[feature] = X_mod[feature].clip(lower=0)

    return X_mod
```

```python
strengths = [
    0.00,
    0.05,
    0.10,
    0.20,
    0.30
]

unconstrained_results = []

for strength in strengths:

    X_modified = apply_unconstrained_perturbation(
        X_detected,
        strength=strength,
        random_state=42
    )

    predictions = rf.predict(X_modified)

    detection_rate = (
        predictions == 1
    ).mean()

    evasion_rate = (
        predictions == 0
    ).mean()

    unconstrained_results.append({
        "Strength": strength,
        "DetectionRate": detection_rate,
        "EvasionRate": evasion_rate
    })

unconstrained_results_df = pd.DataFrame(
    unconstrained_results
)

unconstrained_results_df
```

```python
plt.figure(figsize=(7, 4))

plt.plot(
    unconstrained_results_df["Strength"],
    unconstrained_results_df["DetectionRate"],
    marker="o"
)

plt.xlabel("Unconstrained perturbation strength")
plt.ylabel("Detection rate")
plt.title(
    "Random Unconstrained Feature-Space Perturbation"
)

plt.grid(True)
plt.tight_layout()

plt.show()
```
