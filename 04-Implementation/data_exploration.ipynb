```python
from collections import Counter

label_counts = Counter()
file_rows = {}

for file in csv_files:
    print(f"Reading: {file.name}")

    file_count = 0

    for chunk in pd.read_csv(file, usecols=lambda c: c.strip() == "Label",
                             chunksize=100_000):

        chunk.columns = chunk.columns.str.strip()

        counts = chunk["Label"].value_counts()

        label_counts.update(counts.to_dict())
        file_count += len(chunk)

    file_rows[file.name] = file_count

print("\nFinished.")
```

```python
for filename, rows in file_rows.items():
    print(f"{filename}: {rows:,} rows")
```

```python
print("Class distribution:\n")

for label, count in label_counts.most_common():
    print(f"{label}: {count:,}")
```

```python
class_distribution = pd.DataFrame(
    label_counts.items(),
    columns=["Label", "Count"]
)

class_distribution = class_distribution.sort_values(
    by="Count",
    ascending=False
).reset_index(drop=True)

class_distribution
```

```python
data_quality = []

for file in csv_files:
    print(f"Checking: {file.name}")

    df = pd.read_csv(file)
    df.columns = df.columns.str.strip()

    numeric_cols = df.select_dtypes(include=[np.number]).columns

    nan_count = df.isna().sum().sum()
    inf_count = np.isinf(df[numeric_cols]).sum().sum()
    duplicate_count = df.duplicated().sum()

    data_quality.append({
        "File": file.name,
        "Rows": len(df),
        "NaN": nan_count,
        "Infinity": inf_count,
        "Duplicates": duplicate_count
    })

quality_df = pd.DataFrame(data_quality)
quality_df  
```

```python
duplicate_by_label = Counter()

for file in csv_files:
    print(f"Checking duplicates: {file.name}")

    df = pd.read_csv(file)
    df.columns = df.columns.str.strip()

    duplicates = df[df.duplicated(keep="first")]

    counts = duplicates["Label"].value_counts()
    duplicate_by_label.update(counts.to_dict())

    del df

print("\nDuplicate rows by label:\n")

for label, count in duplicate_by_label.most_common():
    print(f"{label}: {count:,}")
```

```python
duplicate_summary = []

for label, total_count in label_counts.items():

    duplicate_count = duplicate_by_label.get(label, 0)

    percentage = (
        duplicate_count / total_count * 100
        if total_count > 0
        else 0
    )

    duplicate_summary.append({
        "Label": label,
        "Total": total_count,
        "Duplicates": duplicate_count,
        "Duplicate_Percent": percentage
    })

duplicate_summary_df = pd.DataFrame(duplicate_summary)

duplicate_summary_df = duplicate_summary_df.sort_values(
    "Duplicate_Percent",
    ascending=False
).reset_index(drop=True)

duplicate_summary_df
```

```python
bad_rows_summary = []

for file in csv_files:
    df = pd.read_csv(file)
    df.columns = df.columns.str.strip()

    numeric_cols = df.select_dtypes(include=[np.number]).columns

    has_nan = df.isna().any(axis=1)
    has_inf = np.isinf(df[numeric_cols]).any(axis=1)

    bad_rows = has_nan | has_inf

    bad_rows_summary.append({
        "File": file.name,
        "Rows": len(df),
        "Rows_with_NaN": has_nan.sum(),
        "Rows_with_Infinity": has_inf.sum(),
        "Rows_with_any_problem": bad_rows.sum()
    })

    del df

bad_rows_df = pd.DataFrame(bad_rows_summary)
bad_rows_df
```
